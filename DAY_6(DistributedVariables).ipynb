```python
mycollection = "My Name Is Diva and I Love Spark".split(" ")
word = spark.sparkContext.parallelize(mycollection,2)
```

```python
supplementData = {
    "name":1000, "Diva":200, "Love": -300, "Spark": 100
}
```

```python
suppBroadcast = spark.sparkContext.broadcast(supplementData)
```

```python
suppBroadcast.value
```

```python
word.map(lambda word: (word, suppBroadcast.value.get(word, 0))).sortBy(lambda wordPair: wordPair[1]).collect()
```

```python
flights = spark.read.parquet("data/flight-data/parquet/2010-summary.parquet")
```

```python
accChina = spark.sparkContext.accumulator(0)
```

```python
def accChinaFunc(flight_row):
    destination = flight_row["DEST_COUNTRY_NAME"]
    origin = flight_row["ORIGIN_COUNTRY_NAME"]
    if destination == "China":
        accChina.add(flight_row["count"])
    if origin == "China":


        accChina.add(flight_row["count"])
```

```python
flights.foreach(lambda flight_row: accChinaFunc(flight_row))
```
